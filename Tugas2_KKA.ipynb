```python
#Latihan 3
import pandas as pd

df = pd.read_csv('data_kantin - data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolomprint
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Tugas Analisis 3:Kolom menu dan terjual memiliki jumlah non-null lebih sedikit dari jumlah baris total,artinya ada data yang kosong pada kolom tersebut sehingga datanya belum lengkap


```python
#Latihan 4
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Tugas Analisis 4:terjual diisi 0 karena masih bisa dianggap tidak ada penjualan.Menu dihapus karena tanpa nama menu,datanya tidak jelas.

```python
#Latihan 5
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Tugas Analisi 5:Setelah drop_duplicates,jumlah baris berkurang dari 5 menjadi 4 karena data duplikat dihapus,dan tipe data harus benar agar analisis dan perhitungan menghasilkan hasil yang akurat.


```python
#Latihan 6
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Tugas Analisis 6:Nasi Goreng menghasilkan total_pendapatan tertinggi Rp276.000.Informasi ini membantu kantin menentukan menu yang paling menguntungkan sehingga stok dan produksi bisa diprioritaskan.
